```python
# Simple Tic-Tac-Toe

def print_board(board):
   # Displays the current Tic-Tac-Toe board in a readable format
    print("\n")
    for i, row in enumerate(board):
        print(" | ".join(row))
        if i < 2:
            print("-" * 9)

def check_winner(board, player):
  # Checks rows, columns, and diagonals for a winning condition

    for i in range(3):
      # Check row
        if all([cell == player for cell in board[i]]):
            return True

      # Check column
        if all([board[j][i] == player for j in range(3)]):
            return True

     # Check diagonals
    if all([board[i][i] == player for i in range(3)]) or all([board[i][2 - i] == player for i in range(3)]):
        return True
    return False

def is_full(board):
   # Checks if the board is full (draw condition)
    return all(cell in ['X', 'O'] for row in board for cell in row)

def get_move(player, board):
   # Prompts player for input and validates move

    while True:
        try:
            move = input(f"Player {player}, enter your move (row and column as '1 1'): ")
            row, col = map(int, move.strip().split())

             # Convert to 0-based index
            row -= 1
            col -= 1

             # Validate range
            if row not in range(3) or col not in range(3):
                print("Invalid coordinates. Please enter values between 1 and 3.")

             # Check if cell is already taken
            elif board[row][col] != ' ':
                print("Cell already taken. Choose a different move.")
            else:
                return row, col
        except ValueError:
            print("Invalid input format. Please enter row and column numbers separated by a space.")

def play_game():
   # Main game loop controlling turns and game flow

    board = [[' ' for _ in range(3)] for _ in range(3)]
    current_player = 'X'

    while True:
        print_board(board)
        row, col = get_move(current_player, board)
        board[row][col] = current_player

        # Check win condition
        if check_winner(board, current_player):
            print_board(board)
            print(f"Player {current_player} wins!")
            break

        # Check draw condition
        elif is_full(board):
            print_board(board)
            print("It's a draw!")
            break

        # Switch player
        else:
            current_player = 'O' if current_player == 'X' else 'X'

if __name__ == "__main__":
    play_game()
```
